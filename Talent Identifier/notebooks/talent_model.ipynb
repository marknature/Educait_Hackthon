```python
# Connect to SQL database
conn = sqlite3.connect('../data/raw/database.sqlite')
# Load core datasets
players = pd.read_sql("""
    SELECT player_id, player_name, team, position, birth_date, market_value 
    FROM players 
    WHERE league = 'Premier League'
""", conn)
```

```python
matches = pd.read_sql("""
    SELECT match_id, date, home_team, away_team, home_score, away_score 
    FROM matches 
    WHERE season = 2023
""", conn)
events = pd.read_sql("""
    SELECT event_id, player_id, match_id, event_type, minute, x_coord, y_coord, outcome, end_x, end_y
    FROM events 
    WHERE season = 2023
""", conn)
conn.close()
```
